**Question4.** 
- Create a new column department_clean by stripping extra whitespace from department.


```python
df['department_clean'] = df['department'].str.strip(' ')
df
```

**Question5.** 
- (a) Using full_name column create first_name and last_name columns.
- (b) Use email column to extract just the username (the part before the @ is the username) into a new
column.


```python
name_split = df['full_name'].str.split(n = 1, expand=True)
name_split
```

```python
df['first_name'] = name_split[0]
df
```

```python
df['last_name'] = name_split[1]
df
```

```python
# Here split divides email into 2 columns. and 'str[0]' picks up first part
df['email'].str.split('@').str[0]
```

**Question6.** 
- Using department_clean, group the data and find the average salary and the number of employees in each department. Which department has the highest average salary?

```python
department_summary = df.groupby('department_clean').agg(
    avg_salary = ('salary', 'mean'),
    emp_count = ('salary', 'count')
).round(2)
department_summary
```

```python
department_summary['avg_salary'].idxmax()
```

- it Department has the highest average salary
