## Numeric Quantifier
    * a{3} -> the character `a` three times
    * a{3,5} -> the character `a` three, four or five times
    * a{,3} -> the character `a` zero, one, two or three times
    * a{8,} -> the character `a` eight or more times

```python
"mend", "send", "bend"

# sd,sad,saad,saaad
re.search("sa{,3}d", "My name is saad")
```

## Single Character Quantifier

    * a* -> equivalent to a{0,} zero or more
    * a+ -> equivalent to a{1,} one or more
    * a? -> equivalent to a{0,1} zero or one

```python
# 1000-2999
# here [0-9]{3} means that bracket multiply by 3

pattern = "[12][0-9]{3}"

title[title.str.contains(pattern)]
```

```python
# 1000-2999
# \d = [0-9]

pattern = "[12]\d{3}"

title[title.str.contains(pattern)]
```

```python
# email, e-mail

pattern = "e-{0,1}mail"

title[title.str.contains(pattern)]
```

```python
# email, e-mail

pattern = "(e-?mail)" #() is capture group, in extract, capture group is necessary

title.str.extract(pattern).value_counts(dropna= False)
```

```python
# [pdf], [video], [..]
# \w = [A-Za-z0-9_]

pattern = "\[\w+\]"
title[title.str.contains(pattern)]
```

```python
# [pdf], [video], [..]
# . = any character or special character except newline

pattern = "(\[.+\])"
title.str.extract(pattern).value_counts()
```

```python
# [pdf], [video], [..]
# \w = [A-Za-z0-9_]

pattern = "(\[\w+\])"
title.str.extract(pattern).value_counts()
```

## Negative Clas
    * `Negative Set` -> [^fud] == any character except f, u and d
    * `Negative Digit` -> [^0-9] == \D(any negative digit)
    * `Negative Word` -> \W
    * `Negative Whitespace` -> \S any character except space character

```python
a = "I am learning javascript"
b = "I am learning javapython"
c = "Java is my favorite"
d = "I am learning Java"

pattern = "[Jj]ava[^Ss]"

print(re.search(pattern, b))
```

## Lookarounds
- *positive lookahead -> zzz(?=abc) match zzz only when it is followed by abc
- *negative lookahead -> zzz(?!=abc) match zzz only when it is not followed by abc
- *positive lookbehind -> zzz(?<=abc) match zzz only when it is preceded by abc
- *negative lookbehind -> zzz(?<!=abc) match zzz only when it is not preceded by abc


```python
d = "The New C, Standards worth it"

pattern = r"\b[Cc]\b(?!.)"

print(re.search(pattern, d))
```
