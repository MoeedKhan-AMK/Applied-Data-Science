#### DATA QUERY LANGUAGE 

- SELECT
- WHERE
- FILTER OPTIONS
- CASE WHEN (IF ELSE CONDITION) for sql
- COALESCE
- IF NULL
- ORDER BY
- LIMIT

```python
pd.read_sql_query("SELECT * FROM instructor_db ORDER BY starting_salary", conn)
```

```python
pd.read_sql_query("""SELECT * FROM instructor_db 
                    WHERE performance_score > 7 
                    ORDER BY performance_score ASC""", conn)
```

```python
pd.read_sql_query(""" 
    SELECT 
        *,
    CASE 
        WHEN starting_salary >= 150000 THEN "High-Salary"
        WHEN starting_salary <= 160000 THEN "Avg-Salary"
        ELSE "Low-Salary"
    END AS salary_bracket
    FROM instructor_db
    ORDER BY starting_salary DESC
             
""", conn)
```

```python
pd.read_sql_query("SELECT * FROM instructor_db", conn)
```

```python
pd.read_sql_query("""SELECT *, COALESCE(starting_salary, 0) AS sal_for_calculation FROM instructor_db """, conn)
```

```python
pd.read_sql_query("""SELECT *, IFNULL(hire_date,"date unavailable") AS date FROM instructor_db """, conn)
```
