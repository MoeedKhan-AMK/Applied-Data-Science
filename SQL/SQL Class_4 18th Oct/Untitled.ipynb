```python
import pandas as pd
import sqlite3 as sql
```

```python
conn = sql.connect(":memory:")
```

```python
pd.read_sql_query("SELECT * FROM sqlite_master", conn)
```

### DATA DEFINITION LANGUAGE
- This is table changes

- CREATE TABLE
- ALTER TABLE
- DROP TABLE

```python
# CREATING DATABASE TABLE SCHEME

conn.execute("""CREATE TABLE DATABASE 
            (instructor_id INTEGER PRIMARY KEY,
            full_name VARCHAR(100) NOT NULL,
            dept VARCHAR(50),
            hire_date DATE,
            starting_salary DECIMAL(10,2)
            )""")
```

```python
pd.read_sql_query("SELECT * FROM sqlite_master", conn)
```

```python
conn.execute("""ALTER TABLE DATABASE ADD COLUMN performance_score INTEGER""")
```

```python
pd.read_sql_query("SELECT * FROM DATABASE", conn)
```

```python
conn.execute("""ALTER TABLE DATABASE RENAME COLUMN dept TO department""")
```

```python
conn.execute("""ALTER TABLE DATABASE RENAME TO instructor_db""")
```

```python
pd.read_sql_query("SELECT * FROM sqlite_master", conn)
```

```python
conn.execute("""ALTER TABLE instructor_db DROP COLUMN department""")
```

```python
pd.read_sql_query("SELECT * FROM instructor_db", conn)
```
