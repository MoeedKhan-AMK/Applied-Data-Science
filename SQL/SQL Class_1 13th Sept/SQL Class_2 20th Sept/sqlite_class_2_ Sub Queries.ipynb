- `orderby`

```python
# Filtering: imagine a condition where u need to filter, ShareWomen should be greater than 0.5 or  ShareWomen should be less than or equal to 0.5 and Major == "Engineering"
q = """SELECT ShareWomen, Major_Category ,Major FROM recent_grads 
    WHERE ShareWomen > 0.5 or (ShareWomen <= 0.5 and Major_Category == "Engineering") 
    """

pd.read_sql_query(q,conn)
```

```python
# order by ascending order
q = """SELECT ShareWomen, Major_Category ,Major FROM recent_grads 
    WHERE ShareWomen > 0.5 or (ShareWomen <= 0.5 and Major_Category == "Engineering") 
    ORDER BY ShareWomen ASC
    """

pd.read_sql_query(q,conn)
```

```python
# order by descending order
q = """SELECT ShareWomen, Major_Category ,Major FROM recent_grads 
    WHERE ShareWomen > 0.5 or (ShareWomen <= 0.5 and Major_Category == "Engineering") 
    ORDER BY ShareWomen DESC limit 4
    """

pd.read_sql_query(q,conn)
```

```python
# avg function
q = """SELECT avg(ShareWomen) FROM recent_grads 
    WHERE ShareWomen > 0.5 or (ShareWomen <= 0.5 and Major_Category == "Engineering") 
    """

pd.read_sql_query(q,conn)
```

```python
# Limit: Does not work on data agg function

q = """SELECT avg(ShareWomen) FROM recent_grads 
    WHERE ShareWomen > 0.5 or (ShareWomen <= 0.5 and Major_Category == "Engineering") 
    limit 10
    """

pd.read_sql_query(q,conn)
```

```python
# Count function
q = """SELECT avg(ShareWomen), count(Major_Category) FROM recent_grads 
    WHERE ShareWomen > 0.5 or (ShareWomen <= 0.5 and Major_Category == "Engineering") 
    """

pd.read_sql_query(q,conn)
```

- We want to identify unique categories from *Major_Category*

- `GROUP BY` function

```python
q = """SELECT Major_Category, avg(ShareWomen) FROM recent_grads GROUP BY  Major_Category"""

pd.read_sql_query(q,conn)
```

- `HAVING`: The SQL HAVING clause is used to filter grouped data and aggregate functions after a GROUP BY clause has been applied.

```python
q = """SELECT Major_Category, avg(ShareWomen) FROM recent_grads GROUP BY  Major_Category HAVING avg(ShareWomen) > 0.5"""

pd.read_sql_query(q,conn)
```

**SQL alias**
- An SQL alias is a temporary name assigned to a table or column within an SQL query. It is created using the AS keyword and exists only for the duration of that specific query

```python
q = """SELECT Major_Category, avg(ShareWomen) as Average FROM recent_grads GROUP BY  Major_Category HAVING Average > 0.5"""

pd.read_sql_query(q,conn)
```

`cast():` The CAST() function in SQL explicitly converts an expression or column from its current data type into a specified target data type

```python
q = """Select CAST(Women as float) FROM recent_grads"""

pd.read_sql_query(q,conn)
```

```python
q = """SELECT Women, Total FROM recent_grads """

pd.read_sql_query(q,conn)
```

### SUBQUERY in SQL
**Subquery is a query inside a query**

- Imagine a condition where we need only values greater than average of `'ShareWomen'`


```python
q = """SELECT Major, Major_category, ShareWomen FROM recent_grads """

pd.read_sql_query(q,conn)
```

```python
q = """ SELECT avg(ShareWomen) FROM recent_grads"""


pd.read_sql_query(q,conn)
```

- using `Subquery` to meet the condition.
  We will put the `above average query` inside the `query of Major,	Major_category,	ShareWomen rows` 

```python
q = """SELECT Major, Major_category, ShareWomen FROM recent_grads 
        WHERE ShareWomen > (SELECT avg(ShareWomen) FROM recent_grads)"""

pd.read_sql_query(q,conn)
```

- Using `ORDER BY` for columns-wise selection for applying condition.
    - In simple terms it means `ORDER BY 3` means kay 3rd column(ShareWomen) kay behalf pr DESC kro

```python
q = """SELECT Major, Major_Category, ShareWomen FROM recent_grads 
        WHERE ShareWomen > (SELECT avg(ShareWomen) FROM recent_grads)
        ORDER BY 3 DESC
        LIMIT 5"""

pd.read_sql_query(q,conn)
```

```python
q = """SELECT Major, Major_Category, ShareWomen FROM recent_grads 
        WHERE ShareWomen > (SELECT avg(ShareWomen) FROM recent_grads)
        ORDER BY 2 DESC
        LIMIT 5"""

pd.read_sql_query(q,conn)
```
