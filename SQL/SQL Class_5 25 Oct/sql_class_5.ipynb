####  STRFTIME FORMAT

```python
q = """
    SELECT order_id, order_date, 
    STRFTIME ("%Y", order_date) AS order_year,
    STRFTIME ("%m", order_date) AS order_month
    FROM orders
    LIMIT 10
    """

pd.read_sql_query(q, conn)
```

```python
# IF we want to calculate days duration since order
q = """
    SELECT order_id, order_date, 
    CAST(
        JULIANDAY('now') - JULIANDAY(order_date)
    AS INTEGER) AS days_since_order
    FROM orders
    LIMIT 10
    """

pd.read_sql_query(q, conn)
```

```python
# IF we want to calculate days duration since order
q = """
    SELECT STRFTIME("%Y- %m", order_date) AS order_month,
    COUNT(*) AS total_order,
    SUM(
    CASE WHEN status = "Completed" THEN 1 ELSE 0 END) AS completed_orders,
    SUM(
    CASE WHEN status = "Cancelled" THEN 1 ELSE 0 END) AS cancelled_orders
    FROM orders
    GROUP BY STRFTIME("%Y- %m", order_date)
    ORDER BY order_month
    """

pd.read_sql_query(q, conn)
```

#### CT's IN SQL

```python
q = """ 
    SELECT o.order_id, o.customer_id, o.status,
    SUM(oi.quantity * oi.unit_price) AS order_value
    FROM orders o
    JOIN order_items oi
    ON o.order_id = oi.order_id
    GROUP BY o.order_id, customer_id, status 
    
    """

pd.read_sql_query(q, conn)
```

```python
q = """
    WITH order_totals AS (
    SELECT o.order_id, o.customer_id, o.status,
    SUM(oi.quantity * oi.unit_price) AS order_value
    FROM orders o
    JOIN order_items oi
        ON o.order_id = oi.order_id
    GROUP BY o.order_id, customer_id, status),
    
    customer_spending AS (
    SELECT customer_id, SUM(order_value) AS total_spend 
    FROM order_totals
    WHERE status = "Completed"
    GROUP BY customer_id )

    SELECT c.customer_id, c.customer_name, c.city,
    ROUND(cs.total_spend, 2) AS total_spend
    FROM customer_spending AS cs
    JOIN customers AS c

    ON c.customer_id = cs.customer_id
    WHERE cs.total_spend > 200000
    ORDER BY cs.total_spend DESC
    """

pd.read_sql_query(q, conn)
```
