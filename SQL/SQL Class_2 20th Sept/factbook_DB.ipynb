```python
# Here we have used `INNER JOIN` to join two different tables of 'facts and Cities'
# `INNER JOIN`: Returns only rows that have matching values in both tables

query = """SELECT c.*, f.* FROM facts f
        INNER JOIN cities c 
        ON f.id = c.facts_id"""

pd.read_sql_query(query, conn)
```

```python
# Here we renamed columns from two different tables
query = """SELECT c.facts_id, f.id,
        f.name Countries, c.name Cities, c.capital Capital, f.population country_pop, c.population cities_pop 
        FROM facts f
        INNER JOIN cities c 
        ON f.id = c.facts_id"""

pd.read_sql_query(query, conn)
```

```python
# LEFT JOIN: Returns all rows from the left table, and only the matched rows from the right table

query = """SELECT
        f.name Countries, c.name Cities
        FROM facts f
        LEFT JOIN cities c 
        ON f.id = c.facts_id"""

pd.read_sql_query(query, conn)
```

```python
query = """SELECT count(*)
        FROM facts f"""
pd.read_sql_query(query, conn)
```

```python
# Using `WHERE` clause and `IS NULL`  with `LEFT JOIN`:

query = """SELECT
        f.name Countries, c.name Cities
        FROM facts f
        LEFT JOIN cities c 
        ON f.id = c.facts_id
        WHERE Cities IS NULL"""

pd.read_sql_query(query, conn)
```

```python
# Using `WHERE` clause and `IS NOT NULL`  with `LEFT JOIN`:

query = """SELECT
        f.name Countries, c.name Cities
        FROM facts f
        LEFT JOIN cities c 
        ON f.id = c.facts_id
        WHERE Cities IS NOT NULL"""

pd.read_sql_query(query, conn)
```

```python
# RIGHT JOIN: Returns all rows from the right table, and only the matched rows from the left table

query = """SELECT
        f.name Countries, c.name Cities
        FROM facts f
        RIGHT JOIN cities c 
        ON f.id = c.facts_id"""

pd.read_sql_query(query, conn)
```

```python
# FULL JOIN: Returns all rows when there is a match in either the left or right table


query = """SELECT
        f.name Countries, c.name Cities
        FROM facts f
        FULL JOIN cities c 
        ON f.id = c.facts_id"""

pd.read_sql_query(query, conn)
```

##### Find `country, tot_pop, urban_pop, urban_pop_pct`

```python
query = """SELECT f.name Country, f.population Total_Population, SUM(c.population) Urban_population,

        (CAST(SUM(c.population) AS FLOAT) / f.population) * 100 Urban_Population_pct
        
        FROM facts f
        INNER JOIN cities c ON f.id = c.facts_id
        GROUP BY f.id, f.name, f.population"""

pd.read_sql_query(query, conn)
```
