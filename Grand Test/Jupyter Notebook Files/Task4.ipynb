## Task 4:

### Part A – Merge
1. Merge employees.csv and salary.csv using EmployeeID.
2. Merge the resulting dataframe with performance.csv.
3. Display the final dataframe containing:
    - EmployeeID
    - Name
    - Department
    - Salary
    - Rating 


```python
import pandas as pd
```

```python
emp = pd.read_csv("../Test Material/employees.csv")
salary = pd.read_csv("../Test Material/salary.csv")
performance = pd.read_csv("../Test Material/performance.csv")
```

```python
emp
```

```python
salary
```

```python
performance
```

```python
df = pd.merge(emp,salary,on='EmployeeID')
df
```

```python
final_df = pd.merge(df,performance,on="EmployeeID")
final_df
```

### Part B – Concat
4. Concatenate employees.csv and new_employees.csv to create a single employee
database.
5. Display the total number of employees after concatenation. 

```python
new_emp = pd.read_csv("../Test Material/new_employees.csv")
new_emp
```

```python
emp_db = pd.concat([emp,new_emp], ignore_index=True)
emp_db
```

```python
print("Employees in employees.csv file: ", len(emp))
print("Employees in new_employees.csv file: ", len(new_emp))
print("Employees after concat: ", len(emp_db))
```
