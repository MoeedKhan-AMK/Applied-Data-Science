### Question B1  
*6 marks - Medium*

Every file has untidy column labels: trailing spaces, capitals, spaces between words, brackets, and a `%` sign.

Write **one function** that cleans a single label - strip the ends, drop `(` and `)`, turn `%` into `pct`, turn spaces into underscores, lowercase everything - and apply it to the labels of **all six** DataFrames.

````python
# Basic function to apply string functions
def cleaning_labels(labels):
    labels = labels.strip()
    labels = labels.replace("(","")
    labels = labels.replace(")","")
    labels = labels.replace("%","pct")
    labels = labels.replace(" ","_")
    labels = labels.lower()

    return labels

#converting and saving all dataframes into list called 'list_columns_labels'.
list_columns_labels = [customers,products, regions, returns,transactions23, transactions24]

#using for loop and applying to each label
for label in list_columns_labels:
    label.columns = [cleaning_labels(col) for col in label.columns]
````
