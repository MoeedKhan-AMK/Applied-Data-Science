## REGULAR EXPRESSION LEARNING
- 1. re.search()          → find a pattern in text
- 2. span                 → index position of match
- 3. []                   → character class — any one character from the set
- 4. [msb]                → match m or s or b
- 5. [A-Z], [a-z], [0-9]  → ranges inside character class
- 6. [Pp]ython            → case insensitive matching
- 7. \d                   → shortcut for [0-9]
- 8. \w                   → shortcut for [A-Za-z0-9_]
- 9. {3}                  → exactly 3 times
- 10. {3,5}               → 3 to 5 times
- 11. {,3}                → 0 to 3 times
- 12. {8,}                → 8 or more times
- 13. *                   → 0 or more (a{0,})
- 14. +                   → 1 or more (a{1,})
- 15. ?                   → 0 or 1  (a{0,1})
- 16. ()                  → capture group — used with .extract()
- 17. .                   → any character except newline
- 18. str.contains()      → filter rows matching pattern
- 19. str.extract()       → pull out matched text
- 20. value_counts()      → count extracted results
  21.  Remove ALL non-digit characters == .str.replace(r"\D", "", regex=True)
 **\D** means "anything that is NOT a digit"
 This removes spaces, dashes, brackets, dots — everything

**Task 1 — Find all job titles containing "Manager"**
- Use str.contains() with simple pattern
- Expected: rows where job_title has Manager or manager

```python
pattern = "[Mm]anager"

job_title = df["job_title"]

job_title[job_title.str.contains(pattern)]
```

**Task 2 — Find employees in departments starting with "F" or "I"**

- Use character class []
- Hint: [FI]

```python
pattern = "^[FI]"

# Here, we created new emp that contains 'full_name and department' from df
emp = df[["full_name","department"]]   

# here, we said, in emp, 'department' columns should match pattern
result = emp[emp["department"].str.contains(pattern)]
print(result)
```

**Task 4 — Find phone numbers that use spaces as separator**

- "779 898 3950" → has spaces
- "(736) 569-2108" → has brackets, not just spaces
- Use \d{3}\s\d{3} to find space-separated format

```python
pattern = r"\d{3}\s\d{3}\s\d{4}"

phone_num = df["phone"]
phone_num.str.contains(pattern)
phone_num[phone_num.str.contains(pattern, na=False)]
```

```python
# pattern = r"(\d{3}\s\d{3}\s\d{4})| ((\d{3})\s\d{3}-\s\d{4})| (\d{3}-\s\d{3}-\s\d{4})"
pattern = r"(\(\d{3}\)\s\d{3}-\d{4}) | (\d{3}-\d{3}-\d{4}) | (\d{3}\s\d{3}\s\d{4})"

phone_num = df["phone"]
phone_num.str.extract(pattern)
# phone_num[phone_num.str.extract(pattern)]
```

**Task 5 — Extract area code from phone numbers**

- "779 898 3950"   → "779"
- "(736) 569-2108" → "736"
- "747-956-1611"   → "747"
- Hint: use \d{3} with capture group ()

```python
# pattern = r"(\d{3})\s|\((\d{3})\)|(\d{3})-"

pattern = r"(\d{3})"
result = phone_num.str.extract(pattern)

print(result)


# More precise — handle all 3 formats:
# pattern = r"\(?(\d{3})\)?"
# result = phone_num.str.extract(pattern)
# print(result)
```

**Task 7 — Extract year from hire_date**

- "11/08/2021" → "2021"
- "2019-08-02" → "2019"
- "8/6/2018"   → "2018"

```python
hiring = df["hire_date"]
hiring

pattern = r"(\d{4})"

result = hiring.str.extract(pattern)

print(result)
```

**Task 8 — Find phone numbers that have exactly 10 digits**

- First remove all non-digit characters
- Then check if what remains is exactly 10 digits

```python
# cleaning non digits character using == \D

num_clean = phone_num.str.replace(r"\D", "", regex=True)

# finding 10 digits phone numbers
pattern = r"\d{10}$"
result = phone_num[num_clean.str.match(pattern, na=False)]

print(result)
```
