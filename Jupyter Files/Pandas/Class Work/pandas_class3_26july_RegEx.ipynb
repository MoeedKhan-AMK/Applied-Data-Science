## REGULAR EXPRESSION LEARNING
- 1. re.search()          → find a pattern in text
- 2. span                 → index position of match
- 3. []                   → character class — any one character from the set
- 4. [msb]                → match m or s or b
- 5. [A-Z], [a-z], [0-9]  → ranges inside character class
- 6. [Pp]ython            → case insensitive matching
- 7. \d                   → shortcut for [0-9]
- 8. \w                   → shortcut for [A-Za-z0-9_]
- 9. {3}                  → exactly 3 times
- 10. {3,5}               → 3 to 5 times
- 11. {,3}                → 0 to 3 times
- 12. {8,}                → 8 or more times
- 13. *                   → 0 or more (a{0,})
- 14. +                   → 1 or more (a{1,})
- 15. ?                   → 0 or 1  (a{0,1})
- 16. ()                  → capture group — used with .extract()
- 17. .                   → any character except newline
- 18. str.contains()      → filter rows matching pattern
- 19. str.extract()       → pull out matched text
- 20. value_counts()      → count extracted results

## Numeric Quantifier
    * a{3} -> the character `a` three times
    * a{3,5} -> the character `a` three, four or five times
    * a{,3} -> the character `a` zero, one, two or three times
    * a{8,} -> the character `a` eight or more times

```python
"mend", "send", "bend"

# sd,sad,saad,saaad
re.search("sa{,3}d", "My name is saad")
```

## Single Character Quantifier

    * a* -> equivalent to a{0,} zero or more
    * a+ -> equivalent to a{1,} one or more
    * a? -> equivalent to a{0,1} zero or one

```python
# 1000-2999
# here [0-9]{3} means that bracket multiply by 3

pattern = "[12][0-9]{3}"

title[title.str.contains(pattern)]
```

```python
# 1000-2999
# \d = [0-9]

pattern = "[12]\d{3}"

title[title.str.contains(pattern)]
```

```python
# email, e-mail

pattern = "e-{0,1}mail"

title[title.str.contains(pattern)]
```

```python
# email, e-mail

pattern = "(e-?mail)" #() is capture group, in extract, capture group is necessary

title.str.extract(pattern).value_counts(dropna= False)
```

```python
# [pdf], [video], [..]
# \w = [A-Za-z0-9_]

pattern = "\[\w+\]"
title[title.str.contains(pattern)]
```

```python
# [pdf], [video], [..]
# . = any character or special character except newline

pattern = "(\[.+\])"
title.str.extract(pattern).value_counts()
```

```python
# [pdf], [video], [..]
# \w = [A-Za-z0-9_]

pattern = "(\[\w+\])"
title.str.extract(pattern).value_counts()
```
