```python
laptops = pd.read_csv("../Dataset/laptops.csv", encoding="latin-1")
laptops
```

- We want to remove " from ["Screen Size"] Column and wanted to convert its data type to float

```python
laptops["Screen Size"] = laptops["Screen Size"].str.strip('"').astype(float)
laptops.info()
```

- We want to count GPU and CPU models and wanted to save them in new column

```python
laptops["GPU"].value_counts()
```

```python
# we split the gpu column into individual values

gpu_split = laptops["GPU"].str.split(n = 1, expand=True)
gpu_split
```

```python
#saving
laptops['GPU'] = gpu_split[0]
```

```python
laptops['GPU_Model'] = gpu_split[1]
```

```python
laptops["GPU"].value_counts()
```
