## FREQUENCY TABLE IN PYTHON

- Number of times each value appear in table = frequency

- 

- 

- **Syntax:**
            frequency = {}

            for item in data:
                if item in frequency:
                frequency[item] +=1
            else:
            frequency[item] = 1

```python
data = ['Num 10','Num 10', 'Num 28','Num 4', 'Num 15', 'Num 18',  'Num 8', 'Num 7', 'Num 8', 'Num 7','Num 4', 'Num 18', 'Num 18', 'Num 18',  'Num 28', 'Num 28', 'Num 8', 'Num 7' ]
data
```

```python
# Creating Frequency Table
frq_table = {}
for item in data:
    if item in frq_table :
        frq_table[item] += 1
    else:
        frq_table[item] = 1

frq_table #Will return number of times each item appeared in list
```
