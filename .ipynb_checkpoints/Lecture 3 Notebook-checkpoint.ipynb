## LIST in Python
- a built-in data structure 
- store an ordered, changeable collection of items in a single variable

```python
list = ["name",18,28.5,True]
print(list)
```

```python
print(type(list))
```

```python
emp_name = "Khan"
emp_dept = "DS"
emp_salry = 237000
emp_age = 21.8
```

```python
emp_data = [emp_name, emp_dept, emp_salry, emp_age]
print(emp_data)
```

```python
emp2_name = "moares"
emp2_dept = "design"
emp2_salry = 235000
emp2_age = 20.5
```

```python
emp2_data = [emp2_name,emp2_dept,emp2_salry,emp2_age]
print(emp2_data)
```

### INDEXING 

```python
# Suppose we want only salary data of employee from the list: use Indexing
print(emp_data[2])
```

### LIST METHOD: LIST.APPEND()

```python
emp_data.append("Male")
print(emp_data)
```

```python
emp2_data.append("Female")
print(emp2_data)
```

```python
print(emp_data[-3])
```

```python
print(emp2_data[1:3])
```

```python
print(emp_data[:3])
```

```python
print(emp2_data[1:])
```

### EDITING LIST

```python
emp_data[0] = "Moeed Khan"
print(emp_data)
```

### LIST METHOD: LIST.POP()

```python
emp_data.pop(2)
print(emp_data)
```
