## LEGAL VARIABLE in Python Rule:
1. Must Begin with letter(a-z) or _ (underscore)
2. Must not begin with number otherwise error:
   - 1_position❌,
   - position_1✅

```python
_ = "name"
print(_)
```

```python
position_10 = 94
print(position_10)
```

## SPECIAL ASSIGNMENT OPERATOR: +=, -=, /=, *=

```python
value = 99
value+=10
print(value)
```

```python
value = 99
value-=12
print(value)
```

```python
value = 99
value/=9
print(value)
```
