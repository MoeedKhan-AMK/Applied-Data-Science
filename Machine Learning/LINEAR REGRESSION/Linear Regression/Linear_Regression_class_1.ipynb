```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set()

# TO AVOID ELLIPSES BETWEEN COLUMN NAMES
pd.options.display.max_columns = None
pd.options.display.max_rows = None
```

```python
cars = pd.read_csv("./Dataset/CarPrice.csv")
```

```python
cars.head(5)
```

```python
cars.info()
```

```python
cars["CarName"]
```

```python
#Here, split function makes a list from carnames, e.g[alfa-romero, Quadrifoglio],[audi, 100, ls]....

#then we did indexing to fetch name of brand from index [0] and saved that into 'company_name'

company_names = cars["CarName"].apply(lambda x: x.split(" ")[0])
company_names
```

### INSERT() method
- The insert() method allows us to insert a new column to an existing DataFrame.
- Syntax: `df.insert(location, name_for_column, value, allow_duplicates[True/False])`

```python
# to insert as a new column on specific position we use 'df.insert()'
cars.insert(3, "company_name", company_names)
cars
```

```python
cars.drop(["CarName"], axis=1, inplace=True)
```

```python
cars.company_name = cars.company_name.str.lower()
cars.head()
```

```python
cars.company_name.unique()
```

- Multiple duplicate values of `company names` with wrong spelling.
- Writing a function that will take two parameters `wrong name`,`correct name` 

```python
def replace_company_name(wrong_name, correct_name):
    cars.company_name.replace(wrong_name, correct_name, inplace=True)
```

```python
replace_company_name("alfa-romero'","alfa-romeo")
```

```python
replace_company_name("maxda","mazda")
```

```python
replace_company_name("porcshce","porsche")
```

```python
replace_company_name("toyouta","toyota")
```

```python
replace_company_name("vokswagen","volkswagen")
```

```python
replace_company_name("vw","volkswagen")
```

#### PLOTTING GRAPHS
- Plotting Graphs for Price Using `Distribution plot` and `Boxplot`

```python
#giving size of graph
plt.figure(figsize=(20,8))

# plotting for car price distribution as 1st graph
plt.subplot(1,2,1)   
plt.title("Car Price Distribution Plot")
sns.histplot(cars.price, kde=True)


# plotting for car price boxplot as 2nd graph
plt.subplot(1,2,2)
plt.title("Car Price Spread")
sns.boxplot(cars.price)

plt.show()
```
